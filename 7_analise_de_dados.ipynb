```sql
SELECT AVG(idade_cliente) AS media_idade FROM mvp1.gold.netflix_customer_churn;
```

```sql
SELECT 
  genero_cliente,
  COUNT(*) AS quantidade,
  ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER(), 2) AS percentual
FROM mvp1.gold.netflix_customer_churn
GROUP BY genero_cliente
ORDER BY percentual DESC;
```

```sql
SELECT 
  tipo_assinatura,
  COUNT(*) AS total_contratacoes,
  ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER(), 2) AS percentual
FROM mvp1.gold.netflix_customer_churn
GROUP BY tipo_assinatura
ORDER BY total_contratacoes DESC
LIMIT 1;
```

```sql
SELECT 
  dispositivo,
  COUNT(*) AS total_acessos,
  ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER(), 2) AS percentual
FROM mvp1.gold.netflix_customer_churn
GROUP BY dispositivo
ORDER BY total_acessos DESC
LIMIT 1;
```

```sql
SELECT 
  regiao,
  COUNT(*) AS total_consumo,
  ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER(), 2) AS percentual
FROM mvp1.gold.netflix_customer_churn
GROUP BY regiao
ORDER BY total_consumo DESC;

```
