```sql
%python
from pyspark.sql import functions as F

# 1. Carrega a tabela completa sem limite de linhas
df = spark.table("mvp1.gold.netflix_customer_churn")

# 2. Conta quantos nulos existem em cada coluna
contagem_nulos = df.select([F.sum(F.col(c).isNull().cast("int")).alias(c) for c in df.columns])

# 3. Exibe o relatório de nulos na tela
print("Quantidade de valores nulos por coluna na netflix_customer_churn (Tabela Completa):")
contagem_nulos.show()

# 4. Validação do veredito final
total_nulos = sum(contagem_nulos.first().asDict().values())

if total_nulos > 0:
    print(f"❌ TESTE REPROVADO: Foram encontradas {total_nulos} células nulas em toda a tabela.")
else:
    print("✅ TESTE APROVADO: A tabela está 100% preenchida (zero nulos).")
```
