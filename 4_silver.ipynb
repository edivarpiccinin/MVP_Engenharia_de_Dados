```sql
%python
spark.sql("use catalog mvp1")
spark.sql("use schema silver")
```

```sql
drop table if exists netflix_customer_churn
```

```sql
create table mvp1.silver.netflix_customer_churn as
select*
from mvp1.bronze.netflix_customer_churn;
```

```sql
select * from mvp1.silver.netflix_customer_churn
limit 10
```

```sql
alter table mvp1.silver.netflix_customer_churn
set tblproperties ('delta.columnMapping.mode' = 'name')
```

```sql
alter table mvp1.silver.netflix_customer_churn
drop column customer_id;

select * from mvp1.silver.netflix_customer_churn
limit 10;
```

```sql
alter table mvp1.silver.netflix_customer_churn rename column `region` to regiao;
alter table mvp1.silver.netflix_customer_churn rename column device to dispositivo;
alter table mvp1.silver.netflix_customer_churn rename column monthly_fee to taxa_mensal;
alter table mvp1.silver.netflix_customer_churn rename column churned to cancelou_assinatura;
alter table mvp1.silver.netflix_customer_churn rename column payment_method to metodo_pagamento;
alter table mvp1.silver.netflix_customer_churn rename column number_of_profiles to quantidade_perfis;   
alter table mvp1.silver.netflix_customer_churn rename column avg_watch_time_per_day to tempo_medio_assistido;
alter table mvp1.silver.netflix_customer_churn rename column favorite_genre to genero_preferido;

select * from mvp1.silver.netflix_customer_churn
limit 10
```

```sql
update mvp1.silver.netflix_customer_churn
set genero_cliente = case lower (genero_cliente) 
    when 'female' then 'mulher'
    when 'male' then 'homem' 
    when 'other' then 'outro'
    else genero_cliente
end;

select * from mvp1.silver.netflix_customer_churn
limit 10
```
